```python
import pandas as pd
import csv
import requests
from io import StringIO
```

```python
url = "https://aplicacoes.mds.gov.br/sagi/servicos/misocial/"
params = {
    "fq": ["anomes_s:2026*", "valor_repassado_bolsa_familia_s:*"],
    "fl": "codigo_ibge,anomes_s,qtd_familias_beneficiarias_bolsa_familia_s,valor_repassado_bolsa_familia_s,pbf_vlr_medio_benef_f",
    "q": "*:*",
    "rows": 100000,
    "sort": "anomes_s desc, codigo_ibge asc",
    "wt": "csv"
}
```

```python
response = requests.get(url, params=params)
```

```python
response.raise_for_status()
```

```python
csv_data = StringIO(response.text)
reader = csv.DictReader(csv_data)
```

```python
for i, row in enumerate(reader):
    if i< 10:
        print(row)
```

```python
dados_pbf_2026 = pd.read_csv(StringIO(response.text))
```
